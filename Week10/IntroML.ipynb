### Simple Explanation

#### ✅ Labeled Data:

Labeled data means **each input data point has a correct output (label)** already given.

Think of it like this:

> You're learning with a teacher who tells you the correct answer after every question.

📦 Example:

| Image        | Label |
| ------------ | ----- |
| 🐶 Dog image | "Dog" |
| 🐱 Cat image | "Cat" |
| 🚗 Car image | "Car" |

Used in **Supervised Learning**
🔍 Algorithms: Linear Regression, SVM, KNN, Decision Trees, etc.

---

#### ❓ Unlabeled Data:

Unlabeled data means **we only have input data**, and the machine tries to **find patterns** or group things **by itself**.

📦 Example:

| Image        | Label |
| ------------ | ----- |
| 🐶 Dog image | ?     |
| 🐱 Cat image | ?     |
| 🚗 Car image | ?     |

Used in **Unsupervised Learning**
🔍 Algorithms: K-Means, PCA, DBSCAN, etc.

---

### 🎮 Reinforcement Learning (RL)?

Reinforcement Learning is like **learning by playing a game**. An **agent** learns to take actions in an **environment** to **maximize a reward**

📌 Everyday Example:

> A robot learns to walk by trying random steps. If it falls, it gets a negative score. If it moves forward, it gets a reward.

---

### 🔍 Summary Table

| Type          | Data Needed                  | Example Algorithm           |
| ------------- | ---------------------------- | --------------------------- |
| Supervised    | Labeled                      | KNN, SVM, Linear Regression |
| Unsupervised  | Unlabeled                    | K-Means, PCA                |
| Reinforcement | No data; learns from rewards | Q-Learning, SARSA           |

**SUPERVISED LEARNING**

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris, make_regression
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score, mean_squared_error

# -----------------------------
# LINEAR REGRESSION (Supervised)
# -----------------------------
print("\n=== Linear Regression ===")
X, y = make_regression(n_samples=100, n_features=1, noise=10, random_state=1)
model_lr = LinearRegression()
model_lr.fit(X, y)
y_pred = model_lr.predict(X)

plt.figure()
plt.scatter(X, y, color='blue', label='Actual')
plt.plot(X, y_pred, color='red', label='Predicted Line')
plt.title("Linear Regression")
plt.xlabel("Feature")
plt.ylabel("Target")
plt.legend()
plt.grid(True)
plt.show()
print(f"Mean Squared Error: {mean_squared_error(y, y_pred):.2f}")

# -----------------------------
# SUPPORT VECTOR MACHINE (Supervised)
# -----------------------------
print("\n=== Support Vector Machine (SVM) ===")
iris = load_iris()
X = iris.data[:, :2]  # Using only first 2 features for visualization
y = iris.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

model_svm = SVC(kernel='linear')
model_svm.fit(X_train, y_train)
y_pred = model_svm.predict(X_test)

print(f"SVM Accuracy: {accuracy_score(y_test, y_pred)*100:.2f}%")

# Plotting decision boundaries (rough 2D representation)
plt.figure()
for i, color in zip(range(3), ['red', 'blue', 'green']):
    plt.scatter(X[y == i, 0], X[y == i, 1], label=iris.target_names[i], c=color)
plt.title("SVM (2D View of Iris Data)")
plt.xlabel("Sepal length")
plt.ylabel("Sepal width")
plt.legend()
plt.grid(True)
plt.show()

# -----------------------------
# K-NEAREST NEIGHBORS (Supervised)
# -----------------------------
print("\n=== K-Nearest Neighbors (KNN) ===")
model_knn = KNeighborsClassifier(n_neighbors=3)
model_knn.fit(X_train, y_train)
y_pred = model_knn.predict(X_test)
print(f"KNN Accuracy: {accuracy_score(y_test, y_pred)*100:.2f}%")


```

**UNSUPERVISED LEARNING**

```python

# -----------------------------
# K-MEANS CLUSTERING (Unsupervised)
# -----------------------------
print("\n=== K-Means Clustering ===")
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)
clusters = kmeans.labels_

plt.figure()
plt.scatter(X[:, 0], X[:, 1], c=clusters, cmap='viridis')
plt.title("K-Means Clustering")
plt.xlabel("Sepal length")
plt.ylabel("Sepal width")
plt.grid(True)
plt.show()

# -----------------------------
# PRINCIPAL COMPONENT ANALYSIS (PCA - Unsupervised for Dimensionality Reduction)
# -----------------------------
print("\n=== Principal Component Analysis (PCA) ===")
X = iris.data  # Using all 4 features
pca = PCA(n_components=2)
X_reduced = pca.fit_transform(X)

plt.figure()
plt.scatter(X_reduced[:, 0], X_reduced[:, 1], c=iris.target, cmap='rainbow')
plt.title("PCA on Iris Dataset")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.grid(True)
plt.show()
```

**RE-INFORCEMENT LEARNING**

```python
### ✅ Simple Reinforcement Learning Example in Python (Grid Game)
# Very Basic Reinforcement Learning (no libraries)
# Grid: 0 = empty, 1 = goal
# Agent starts at 0,0 and learns to move right to goal

import random

# Define environment
grid = [[0, 0, 0, 1]]
actions = ['right', 'left']  # limited actions
state = 0  # starting at position 0
goal_state = 3
rewards = [0, 0, 0, 10]  # goal gives +10 reward
q_values = [0, 0, 0, 0]  # initially all zero

alpha = 0.1  # learning rate
gamma = 0.9  # discount factor
episodes = 50

for episode in range(episodes):
    state = 0
    done = False
    while not done:
        action = random.choice(actions)

        if action == 'right':
            next_state = min(state + 1, 3)
        else:
            next_state = max(state - 1, 0)

        reward = rewards[next_state]
        q_values[state] = q_values[state] + alpha * (reward + gamma * q_values[next_state] - q_values[state])

        state = next_state
        if state == goal_state:
            done = True

print("Final Q-values (what agent learned):", q_values)
```

```python

# ============================================
# 🎮 Simple Reinforcement Learning Example (Q-Learning)
# No libraries required, fully beginner-friendly!
# Goal: Move the agent to the last cell (goal) and learn best actions
# ============================================

import random
import time

# Define environment
# Grid: [0, 0, 0, 1] (last cell is the goal with reward)
grid = [0, 0, 0, 1]
actions = ['right', 'left']
rewards = [0, 0, 0, 10]  # Only the goal gives reward

# Initialize Q-values (what the agent learns)
q_values = [0, 0, 0, 0]

# Parameters
alpha = 0.1       # Learning rate
gamma = 0.9       # Discount factor
episodes = 30     # Number of training episodes

def show_grid(state):
    display = ['🤖' if i == state else '⬜' for i in range(4)]
    display[-1] = '🏁'  # Goal
    print(' '.join(display))

print("\n🏁 Training Agent to Reach the Goal...")

for episode in range(episodes):
    state = 0  # Start at beginning
    done = False
    print(f"\nEpisode {episode+1}")
    show_grid(state)
    time.sleep(0.3)

    while not done:
        action = random.choice(actions)

        # Move the agent
        if action == 'right':
            next_state = min(state + 1, 3)
        else:
            next_state = max(state - 1, 0)

        # Get reward and update Q-value
        reward = rewards[next_state]
        q_values[state] = q_values[state] + alpha * (reward + gamma * q_values[next_state] - q_values[state])

        state = next_state
        show_grid(state)
        time.sleep(0.3)

        if state == 3:
            done = True

print("\n✅ Final Q-values (agent's learned value at each state):")
print(q_values)
```
